````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    write = 0

    # Copy nonzero values forward in their original order.
    for value in nums:
        if value != 0:
            nums[write] = value
            write += 1

    # Fill the remaining positions with zeros.
    while write < len(nums):
        nums[write] = 0
        write += 1

    return nums
````

3. Based on the provided starter code, code the solution to your assigned problem here in Python. Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Try to find the solution with best time and space complexity.

````python
# 
from typing import List
def move_zeros_to_end(nums: List[int]) -> List[int]:
    write = 0

    # Move each nonzero number to the next available position.
    for value in nums:
        if value != 0:
            nums[write] = value
            write += 1

    # Set all remaining positions to zero.
    while write < len(nums):
        nums[write] = 0
        write += 1

    return nums
````


4. Walk through your Python code in plain words and explain why your solution works.


````python
# The variable write marks the next position where a nonzero number should go. The for loop reads the list from left to right. Whenever it finds a nonzero number, it writes that number at nums[write] and advances write. Since the numbers are read in their original order, they are also written in that order. After the loop, every position before write contains a nonzero number. The while loop fills all remaining positions with zeros. This puts every zero at the end while preserving the order of the nonzero numbers.
# For example, with [0, 1, 0, 3, 12], the nonzero numbers are written to the start of the list in order: 1, 3, and 12. The remaining positions are then set to zero, giving [1, 3, 12, 0, 0]
````


5. Explain the time and space complexity of your solution.


````python
# Time: O(n), where n is the number of elements. The function scans the list and then fills the remaining positions.
# Extra space: O(1), because it changes the input list directly and uses only a few variables.
````
