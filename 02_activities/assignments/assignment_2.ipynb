```python
# Solving Assigned problem 3

from typing import List, Union

def missing_num(nums: List[int]) -> Union[List[int], int]:
    present = set(nums)
    missing = []

    for number in range(1, len(nums) + 1):
        if number not in present:
            missing.append(number)

    if not missing:
        return -1

    return missing

# A set is useful here because it stores each number only once and makes membership checks fast. We can compare the numbers in the list with every number in the required range, 1 through n.
# The starter code’s return type says int, but the function can return either a list of missing numbers or -1. This version above uses Union to describe both possibilities
# For nums = [6, 8, 2, 3, 5, 7, 1, 1, 2, 10], the set contains 1, 2, 3, 5, 6, 7, 8, 10. Checking the range from 1 to 10 shows that 4 and 9 are missing, so the function returns [4, 9].
# The duplicate values do not cause a problem: the set records whether each number appears at least once. If every number from 1 to n appears, missing stays empty and the function returns -1.
# Complexity: O(n) time and O(n) extra space. The function scans the input to build the set and checks each number in the range.
```
